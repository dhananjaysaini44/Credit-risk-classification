```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import RobustScaler
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression   
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, precision_score, recall_score   # to measure precision
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("credit_risk_dataset.csv")
```

```python
df.head(5)
```

```python
df["Education_Level"].unique()
```

```python
order = {'Bachelors': 1, 'PhD': 3, 'Masters': 2, 'High School': 0}
```

```python
df["Education_Level"] = df["Education_Level"].map(order)
```

```python
df.head(5)
```

```python
df["Housing_Status"].unique()
```

```python
OneHot = OneHotEncoder(drop = None, sparse_output = False)
HousingOneHotEncoded = OneHot.fit_transform(df[["Housing_Status"]])
HousingOneHotEncodedDf = pd.DataFrame(HousingOneHotEncoded, columns = OneHot.get_feature_names_out(["Housing_Status"]), index = df.index)
df = pd.concat([df, HousingOneHotEncodedDf], axis = 1)
df = df.drop(columns = ["Housing_Status"])
```

```python
df.isnull().sum()
```

```python
df["Income"] = df["Income"].fillna(df["Income"].mean())
```

```python
y = df["Default"]
x = df.drop(columns = ["Default"])
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 12)
```

```python
model = LogisticRegression(max_iter = 1000)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
model.score(x_test, y_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
df["Default"].value_counts()
```

```python
recall_score(y_test, y_pred)
```

```python
precision_score(y_test, y_pred)
```
